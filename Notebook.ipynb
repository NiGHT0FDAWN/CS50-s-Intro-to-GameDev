#### Love2D Important Functions:

.```
- love.load()                       # initalizing the game state
- love.update()                     # called each frame by LOVE, dt is delta time in seconds from one frame to the next *(dt)*
- love.draw()                       # called each frame after update for drawing things to the screen once things have change
- love.graphics.printf()            # versatile print function *(text, x, y, [width], [height])*
- love.window.setMode()             # initializing the window, along with various window parameters like VSync, fuklscreen etc *(width, height, params)*
- love.graphics.setDefaultFilter()  # sets texture scaling filter when scaling textures/fonts. The default is bilinear *(min, mag)*
- love.keypressed()                 # callback function that executes whenever we press a key. *(key)*
- love.event.quit()                 # you guessed it! this terminates the app.
```
