#### Love2D Important Functions:

.```
- love.load()                       # initalizing the game state
- love.update()                     # *(dt)* called each frame by LOVE, [dt] is delta time in seconds from one frame to the next 
- love.draw()                       # called each frame after update for drawing things to the screen once things have change
- love.graphics.printf()            # *(text, x, y, [width], [height])* versatile print function 
- love.window.setMode()             # *(width, height, params)* initializing the window, along with various window parameters like [vsync], [fullscreen], etc 
- love.graphics.setDefaultFilter()  # *(min, mag)* sets texture scaling filter when scaling textures/fonts. The default is [bilinear]
- love.keypressed()                 # *(key)* callback function that executes whenever we press a key. 
- love.event.quit()                 # you guessed it! this terminates the app.
- love.graphics.newFont()           # *(path, size)* loads a font file, sets a specific size, and stores it 
- love.graphics.setFont()           # *(font)* Sets current active [font]
- love.graphics.clear()             # *(r, g, b, a)* fills screen with color by RGBA set ([0]-[255])
- love.graphics.rectangle()         # *(mode, x, y, width, height)* draws a rectangle onto the screen using whichever our active color is. Can be set to mode [fill] or [line]
- love.window.setTitle()            # *(title)* Sets the title of the window
- love.timer.getFPS()               # returns the current FPS of our application
- love.audio.newSource()            # *(path, [type])* Creates a Love2D audio object that we can play back. [type] includes [stream] and [static]
- love.resize()                     # *(width, height)* Called by LOVE2D everytime we resize the app. Logic should go in here if anything in the game is sized dynamically (like UI) based on window size.
```
